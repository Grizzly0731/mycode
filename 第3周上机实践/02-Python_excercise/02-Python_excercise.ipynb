### 1.6. 互联网上的每台计算机都有一个IP，合法的IP格式为：A.B.C.D。其中A、B、C、D均为[0, 255]中的整数。为了简单起见，我们规定这四个整数中不允许有前导零存在，如001。现在给你一个字符串s（s不含空白符），请你判断s是不是合法IP，若是，输出Yes，否则输出No。如：s=“202.115.32.24”，则输出Yes；s=“a.11.11.11”, 则输出No。

```python
s = input().strip()

parts = s.split(".")
ok = len(parts) == 4

if ok:
    for p in parts:
        if p == "":
            ok = False
            break
        for ch in p:
            if ch < "0" or ch > "9":
                ok = False
                break
        if not ok:
            break
        if len(p) > 1 and p[0] == "0":
            ok = False
            break
        if int(p) > 255:
            ok = False
            break

print("Yes" if ok else "No")
```

### 1.7. 创建一个名为sandwich_orders 的列表，在其中包含各种三明治的名字；再创建一个名为finished_sandwiches 的空列表。遍历列表sandwich_orders，对于其中的每种三明治，都打印一条消息，如I made your tuna sandwich，并将其移到列表finished_sandwiches。所有三明治都制作好后，打印一条消息，将这些三明治列出来。

```python
sandwich_orders = ['tuna', 'beef', 'veggie', 'chicken']
finished_sandwiches = []

while sandwich_orders:
    sandwich = sandwich_orders.pop(0)
    print(f"I made your {sandwich} sandwich")
    finished_sandwiches.append(sandwich)

print("\nAll sandwiches are ready:")
for s in finished_sandwiches:
    print(f"  - {s}")
```

### 2.2.1 基础类与对象：创建一个 Person 类，包含属性 name 和 age，以及方法 introduce()，用于打印个人信息。

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"我叫{self.name}，今年{self.age}岁。")

p = Person("小明", 20)
p.introduce()
```

### 2.2.2 继承：创建一个 Student 类，继承自 Person 类，并添加属性 student_id 和方法 study()。

```python
class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)      # 先让父类把 name/age 存好
        self.student_id = student_id     # 再加自己的属性

    def study(self):
        print(f"{self.name}（学号 {self.student_id}）正在学习。")

s = Student("小明", 20, "2024001")
s.introduce()   # 父类的方法，直接能用
s.study()       # 自己的方法
```

### 2.2.4 类方法与静态方法：创建一个 MathUtils 类，包含一个类方法 add() 和一个静态方法 multiply()

```python
class MathUtils:
    @classmethod
    def add(cls, a, b):
        return a + b

    @staticmethod
    def multiply(a, b):
        return a * b

print(MathUtils.add(2, 3))        # 5
print(MathUtils.multiply(2, 3))   # 6
```

### 2.2.5 属性装饰器：创建一个 Circle 类，包含属性 radius，并通过属性装饰器确保半径不能为负数。

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius          # 也会走下面的 setter，负数直接被拦住

    @property
    def radius(self):
        return self._radius           # 真正存值的是 _radius

    @radius.setter
    def radius(self, value):
        if value < 0:
            raise ValueError("半径不能为负数")
        self._radius = value
```

### 2.2.7 魔术方法（特殊函数）：创建一个 Point 类，有x和y两个属性，表示二维坐标系中的点，并实现 __add__ 和 __str__ 魔术方法。

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):                    # 决定 p1 + p2 怎么算
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):                           # 决定 print(p) 显示什么
        return f"({self.x}, {self.y})"


p1 = Point(1, 2)
p2 = Point(3, 4)

print(p1 + p2)     # (4, 6)
print(p1)          # (1, 2)
```

### 2.2.8 单例模式：实现一个单例模式的 Logger 类，确保只有一个实例。

```python
class Logger:
    _instance = None

    def __new__(cls):
        if cls._instance is None:          # 第一次创建才真正 new
            cls._instance = super().__new__(cls)
        return cls._instance               # 以后都返回同一个

    def log(self, msg):
        print(f"[LOG] {msg}")


a = Logger()
b = Logger()
print(a is b)      # True
a.log("hello")     # [LOG] hello
```

### 2.2.9 抽象基类：创建一个抽象基类 Shape，并实现两个子类 Circle 和 Rectangle，要求每个子类都必须实现 area() 方法。

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass            # 只声明，不实现

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14159 * self.r ** 2

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w, self.h = w, h

    def area(self):
        return self.w * self.h


for s in (Circle(2), Rectangle(3, 4)):
    print(round(s.area(), 2))
    
```

### 2.3.1 以列表存储矩阵，编写一个函数，计算两个矩阵的乘积

```python
def matmul(A, B):
    n, m = len(A), len(A[0])      # A: n 行 m 列
    p = len(B[0])                 # B: m 行 p 列
    C = [[0] * p for _ in range(n)]

    for i in range(n):
        for j in range(p):
            total = 0
            for k in range(m):
                total += A[i][k] * B[k][j]   # 第 i 行 × 第 j 列
            C[i][j] = total
    return C
```

## Step 10: 按照船票价格降序排列

```python
# 请填写相应代码
titanic.sort_values("Fare", ascending=False)
```

## Step 11: 绘制一个展示船票价格的直方图

```python
plt.rcParams["font.sans-serif"] = ["Microsoft YaHei", "SimHei"]
plt.rcParams["axes.unicode_minus"] = False
titanic["Fare"].plot(kind="hist", bins=30, edgecolor="black")
plt.xlabel("船票价格")
plt.ylabel("人数")
plt.title("船票价格分布")
plt.show()
```

## Step 12: 绘制乘客年龄与船票价格的散点图

```python
plt.scatter(titanic["Age"], titanic["Fare"], alpha=0.5, s=10)
plt.xlabel("年龄")
plt.ylabel("船票价格")
plt.title("年龄与船票价格的关系")
sns.scatterplot(data=titanic, x="Age", y="Fare", hue="Survived", alpha=0.7)
plt.show()
```

### 2. 生成一个4X5和一个5X4的矩阵，并计算它们的乘积。

```python
import numpy as np
juzhen1=np.random.randint(1,100,size=(4,5))
juzhen2=np.random.randint(1,100,size=(5,4))#随机生成两个矩阵
print(juzhen1,"\n\n",juzhen2)

result=juzhen1@juzhen2#计算

print("\n",result)

```

### 3. 函数 f 定义如下，其中x是任意实数: 𝑓(𝑥)=(sin⁡(𝑥+1)/𝑒^𝑥 )(𝑥+1)/2，生成一个数组a∈𝑅^5，计算b = f(a)。

```python
import numpy as np
def f (x):
    result=(np.sin(x+1)/np.exp(x))*(x+1)/2
    return result
a=np.random.randint(0,100,5)
b=f(a)
print(a,"\n",b)
    
```

### 4. 随机生成一个5行10列的矩阵，然后每行元素减去该行的平均值

```python
import numpy as np
a=np.random.randint(100,size=(5,10))
print(a)
ava=np.mean(a,axis=1,keepdims=True)
b=a-ava
print()
print(np.round(b,2))
```

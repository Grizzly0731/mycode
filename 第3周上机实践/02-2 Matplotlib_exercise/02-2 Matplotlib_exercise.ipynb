## Matplotlib练习

### 1. 用蓝线绘制函数𝑓(𝑥) = 𝑥^2−2𝑥+3的图表，并在坐标(1,2)位置处标上一个红色的点


```python
import matplotlib.pyplot as ma
import numpy as np
x=np.linspace(-20,20,500)
y=x**2-2*x+3
ma.plot(x,y,'b-')
ma.scatter(1,2,color='red',s=100,zorder=3,label='坐标')
ma.show()
```

### 2. 使用np.linspace()函数生成t∈[0, 2𝜋]。给定𝑥=16*sin⁡(𝑡)^3和𝑦=13*cos⁡(𝑡)−5*cos⁡(2𝑡)−2*cos⁡(3𝑡)−cos⁡(4𝑡)，画出x-y图表，并给图标添加一个题目：Heart

```python
import numpy as np
import matplotlib.pyplot as plo
t=np.linspace(0,2*np.pi,500)
x=16*np.sin(t)**3
y=13*np.cos(t)-5*np.cos(2*t)-np.cos(4*t)
plo.plot(x,y,'b-')
plo.title("Heart")
plo.xlabel("x")
plo.ylabel("y")
plo.show()
```

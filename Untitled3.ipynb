```python
import numpy as np, pandas as pd
from collections import Counter

df = pd.read_csv(r"C:\users\asus\Documents\mycode\iris_dataset.csv")
df.head()
```

```python
X, y = df.iloc[:, :4].to_numpy(), df["target_name"].to_numpy()
print(X.shape, y.shape)      # (150, 4) (150,)
```

```python
rng = np.random.default_rng(42)
train, test = [], []
for name in np.unique(y):
    idx = np.where(y == name)[0]
    rng.shuffle(idx)
    train += list(idx[:40]); test += list(idx[40:])

print(len(train), len(test))  # 120 30
```

```python
mu, sd = X[train].mean(0), X[train].std(0)
Xtr, Xte = (X[train] - mu) / sd, (X[test] - mu) / sd
ytr, yte = y[train], y[test]
print(Xtr.shape, Xte.shape)   # (120, 4) (30, 4)
```

```python
def knn(Xa, ya, Xb, k):
    d = ((Xb[:, None] - Xa[None]) ** 2).sum(2)
    return np.array([Counter(ya[r]).most_common(1)[0][0]
                     for r in np.argsort(d, 1)[:, :k]])
```

```python
def cv(k):
    s = 0
    for f in range(5):
        va = np.arange(len(ytr)) % 5 == f
        s += (knn(Xtr[~va], ytr[~va], Xtr[va], k) == ytr[va]).mean()
    return s / 5

ks = list(range(1, 32, 2))
scores = [cv(k) for k in ks]
best_k = ks[int(np.argmax(scores))]
print("最优 K =", best_k, " CV =", round(max(scores), 4))
```

```python
pred = knn(Xtr, ytr, Xte, best_k)
print("准确率 =", round((pred == yte).mean(), 4))
print(pd.crosstab(yte, pred))
```

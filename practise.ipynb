```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pandas as pd
DATA_PATH = r"C:\users\asus\Documents\mycode\iris_dataset.csv"
df = pd.read_csv(DATA_PATH)
print(type(df))          # 确认类型
print("形状:", df.shape)   # (行数, 列数)
print("列名:", list(df.columns))
df.head()                # Jupyter 里会渲染成表格
X = df.drop(columns="target_name")   # 丢掉标签列，剩下全是特征
y = df["target_name"]                # 直接按列名取标签
rng = np.random.default_rng(42)

train_idx, test_idx = [], []
for name in y.unique():                    # 依次处理三个类别
    idx = y[y == name].index.to_numpy().copy()    # 取出这一类的行号
    rng.shuffle(idx)                       # 类内打乱
    cut = int(len(idx) * 0.8)              # 每类 50 条 → 切 40 条
    train_idx.append(idx[:cut])
    test_idx.append(idx[cut:])

train_idx = np.sort(np.concatenate(train_idx))
test_idx  = np.sort(np.concatenate(test_idx))

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
```

```python
score=[89,92,59,70]
subject=['语文','数学','英语','体育']
total=0
ava=0
max_=0
i=0
while i<len(score):
    total+=score[i]
    i+=1
print('总分',total)
ava=total/len(subject)
print ('均值',ava)
i=0
while i<len(score):
    if max_<score[i]:
        max_=score[i]
    i+=1
print ('最大值',max_)
```

```python
def cal_score():
  user_input=input ('请输入各科成绩，用英文逗号隔开：')
  score=[int(x)for x in user_input.split(',')] 
  total=0
  total=sum(score)
  ava=total/len(score)
  max_=max(score)
  print ('总分',total,'均值',ava,'最大值',max_)
cal_score()
 
```
